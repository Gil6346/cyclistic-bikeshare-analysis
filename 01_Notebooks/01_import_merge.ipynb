# Cyclistic Bike Share Analysis

## Notebook 1 - Importing and Merging Data

Objective:
Import all 12 monthly trip datasets and merge

Input: 00_Data/00_Raw/

Output: 00_Data/01_Interim/cyclistic_202506_202605_merged.parquet

```python
# Import libraries

import numpy as np
import pandas as pd
from pathlib import Path
```

```python
# Set data folder

DATA_PATH = Path("../00_Data/00_Raw")

DATA_PATH
```

```python
# Set csv files path and verification

csv_files = sorted(DATA_PATH.glob("*.csv"))

print(len(csv_files))

for files in csv_files:
    print(files.name)
```

```python
# Import 12 csv files

dfs = []
for file in csv_files:
    print(f"Reading {file.name}")
    dfs.append(pd.read_csv(file))
```

```python
# Schema validation

reference_columns = dfs[0].columns

schema_mismatch = False

for i in range(len(dfs)):
    if not dfs[i].columns.equals(reference_columns):
        schema_mismatch = True
        print(f"Schema mismatch found in: {csv_files[i].name}")

if not schema_mismatch:
    print("All schemas are identical.")

# Schema columns difference

reference = set(dfs[0].columns)

for i, df in enumerate(dfs):
    cols = set(dfs[i].columns)

    missing = reference - cols
    extra = cols - reference

    if missing or extra:
        print(f"\n{csv_files[i].name}")
        if missing:
            print("Missing: ", missing)
        if extra:
            print("Extra: ", extra)
```

```python
# Merging 12 datasets into 1 unified dataset

df = pd.concat(dfs, ignore_index=True)
```

```python
# Export merged dataset

OUTPUT_PATH = Path("../00_Data/01_Interim")
OUTPUT_PATH.mkdir(parents=True, exist_ok=True)

df.to_parquet(
    OUTPUT_PATH / "cyclistic_202506_202605_merged.parquet",
    index=False
)
```

```python
# Validation

test = pd.read_parquet(
    OUTPUT_PATH / "cyclistic_202506_202605_merged.parquet"
)

assert test.shape == df.shape

assert list(test.columns) == list(df.columns)
```
